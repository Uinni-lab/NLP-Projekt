# Schritt 3: Ansatz 1: VADER

Vader wurde auf englischsprachigen Social-Media-Texten (Tweets) entwickelt und validiert, nicht auf Songtexten. Man erwartet daher, dass die Übertragung auf Lyrics nicht 1:1 funktioniert. Deswegen auch das zusätzliche Transformer Model.

```python
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
```

```python
def berechne_vader_werte(text):
    werte = analyzer.polarity_scores(text)
    return werte
```

```python
def klassifiziere_compound(compound_wert):
    # Schwellenwerte 0.05 / -0.05 sind die von Vader selbst empfohlenen Standardgrenzen (siehe Hutto & Gilbert 2014)
    if compound_wert >= 0.05:
        return 'positiv'
    elif compound_wert <= -0.05:
        return 'negativ'
    else:
        return 'neutral'
```

```python
def berechne_vader_für_datensatz(daten):
    compound_liste = []
    kategorie_liste = []
    for index, zeile in daten.iterrows():
        text = zeile['lyrics_clean']
        werte = berechne_vader_werte(text)
        compound_liste.append(werte['compound'])
        kategorie_liste.append(klassifiziere_compound(werte['compound']))
    daten['vader_compound'] = compound_liste
    daten['vader_kategorie'] = kategorie_liste
    return daten
```

#Pipeline ausführen

```python
daten = pd.read_csv('data/daten_vorverarbeitet.csv')
daten = berechne_vader_für_datensatz(daten)
print(daten[['title', 'vader_compound', 'vader_kategorie']])
daten.to_csv('data/daten_mit_vader.csv', index=False)
print('Ergebnisse gespeichert unter: data/daten_mit_vader.csv')
```
